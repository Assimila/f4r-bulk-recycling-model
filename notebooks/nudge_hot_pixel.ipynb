This notebook demonstrates how to use `identify_hot_pixel` and `nudge_hot_pixel` after identifying a hot pixel which causes numerical instability

```python
import numpy as np
import xarray as xr
from matplotlib import pyplot as plt
```

```python
ds = xr.open_dataset("dao.80_93.nc")
```

```python
if ds.coords["level"].isnull().all():
    print("Fixing broken pressure level data")
    levels = [1000.0, 950.0, 900.0, 850.0, 700.0, 500.0, 300.0, 200.0]
    ds.coords["level"] = xr.DataArray(levels, dims=["level"], coords={"level": levels}, attrs={"units": "hPa"})
```

```python
# make sure lat runs from south to north
if not ds["lat"].to_index().is_monotonic_increasing:
    print("flipping lat")
    ds = ds.sortby("lat", ascending=True)
```

```python
# make sure lon runs from west to east
if not ds["lon"].to_index().is_monotonic_increasing:
    print("flipping lon")
    ds = ds.sortby("lon", ascending=True)
```

```python
# get a spatial subset 4°N–16°S, 50°–76°W
ds = ds.sel(lat=slice(-16, 4), lon=slice(-76, -50))
```

```python
# make sure that the order of the dimensions is (lon, lat, ...) for all variables
ds = ds.transpose("lon", "lat", ...)
```

```python
# this time step does not converge
# should be Sept 1980
ds = ds.isel(time=8, drop=True)
```

```python
import bulk_recycling_model.numerical_integration
```

```python
# Integrate 10^-3 Shum Uwnd dp
# Because the integration limits are from high pressure to low pressure, we need to invert the sign.
integrand = -1 * 1e-3 * ds["Shum"] * ds["Uwnd"]
Fx = bulk_recycling_model.numerical_integration.integrate_no_extrapolation(integrand, ds["Psfc"])
# Units: mb x m/s
```

```python
# Integrate 10^-3 Shum Vwnd dp
# Because the integration limits are from high pressure to low pressure, we need to invert the sign.
integrand = -1 * 1e-3 * ds["Shum"] * ds["Vwnd"]
Fy = bulk_recycling_model.numerical_integration.integrate_no_extrapolation(integrand, ds["Psfc"])
# Units: mb x m/s
```

Prepare and scale the data

```python
from bulk_recycling_model import plotting, preprocess
from bulk_recycling_model.axis import Axis
from bulk_recycling_model.coefficients import Coefficients
from bulk_recycling_model.scaling import Scaling, UnitSystem
```

```python
# degrees
L = ds.coords["lon"].max().item() - ds.coords["lon"].min().item()
# convert to meters
L = L * 111e3 * np.cos(np.deg2rad(ds.coords["lat"].mean().item()))
dx = L / ds.sizes["lon"]
```

```python
# lon axis
lon_axis = Axis(
    ds.coords["lon"].min().item(),
    ds.coords["lon"].diff("lon").mean().item(),
    ds.sizes["lon"],
)
```

```python
# degrees
H = ds.coords["lat"].values[-1] - ds.coords["lat"].values[0]
# convert to meters
H = H * 111e3
dy = H / ds.sizes["lat"]
```

```python
# lat axis
lat_axis = Axis(
    ds.coords["lat"].min().item(),
    ds.coords["lat"].diff("lat").mean().item(),
    ds.sizes["lat"],
)
```

```python
print(f"{L = :.2e} m")
print(f"{dx = :.2e} m")
print(f"{H = :.2e} m")
print(f"{dy = :.2e} m")
```

```python
# make a scaling object to convert between unit systems
scaling = Scaling(H)
```

```python
dx = scaling.distance.convert(dx, UnitSystem.SI, UnitSystem.scaled)
dy = scaling.distance.convert(dy, UnitSystem.SI, UnitSystem.scaled)
print(f"{dx = :.2e} scaled")
print(f"{dy = :.2e} scaled")
```

```python
# convert Fx and Fy to scaled units
Fx = scaling.water_vapor_flux.convert(Fx.values, UnitSystem.natural, UnitSystem.scaled)
Fy = scaling.water_vapor_flux.convert(Fy.values, UnitSystem.natural, UnitSystem.scaled)
```

```python
# preprocess water vapor fluxes onto the secondary grid
Fx_left = preprocess.prepare_Fx_left(Fx)
Fx_right = preprocess.prepare_Fx_right(Fx)
Fy_bottom = preprocess.prepare_Fy_bottom(Fy)
Fy_top = preprocess.prepare_Fy_top(Fy)
```

```python
# convert E to scaled units
E = scaling.evaporation.convert(ds["Evap"].values, UnitSystem.natural, UnitSystem.scaled)
```

```python
# preprocess E onto the secondary grid
E = preprocess.prepare_E(E)
```

```python
# compute P
P = preprocess.calculate_precipitation(Fx_left, Fx_right, Fy_bottom, Fy_top, E, dx, dy)
```

Plot the instability heuristic

```python
coeffs = Coefficients(Fx_left, Fx_right, Fy_bottom, Fy_top, E, P, dx, dy)
```

```python
fig, ax = plt.subplots()
collection = plotting.pcolormesh(ax, coeffs.instability_heuristic, lon_axis, lat_axis, vmin=0.0)
fig.colorbar(collection, ax=ax, label="instability heuristic")
```

Run the model - demonstrating that the model fails to converge

```python
import logging

logging.basicConfig()
logging.getLogger("bulk_recycling_model").setLevel(logging.INFO)
```

```python
from bulk_recycling_model.main import run
```

```python
status = run(
    Fx_left,
    Fx_right,
    Fy_bottom,
    Fy_top,
    E,
    P,
    dx,
    dy,
    tol=1e-2,
)
```

```python
# does not converge
assert status["success"] is False
```

```python
# plot the final solution
rho = status["rho"]
fig, ax = plt.subplots()
collection = plotting.pcolormesh(ax, rho, lon_axis, lat_axis, vmin=0.0, vmax=0.8)
fig.colorbar(collection)
fig.suptitle("$\\rho$");
```

```python
# plot the convergence
deltas = status["deltas"]
fig, ax = plt.subplots()
ax.plot(deltas)
ax.set_title("Convergence")
ax.set_xlabel("Iteration");
```

identify the hot pixel

```python
from bulk_recycling_model.numerical_stability import identify_hot_pixel, nudge_hot_pixel
```

```python
i_hot, j_hot = identify_hot_pixel(coeffs.instability_heuristic)
print(
    f"Hot pixel identified at (i={i_hot}, j={j_hot}) "
    f"with instability heuristic = {coeffs.instability_heuristic[i_hot, j_hot]}"
)
print("E =", E[i_hot, j_hot])
print("P =", P[i_hot, j_hot])
```

nudge the hot pixel

```python
# tune these values as needed
offset = 3.5
kernel_size = 3
```

```python
E_nudged = nudge_hot_pixel(E, i_hot, j_hot, offset=offset, kernel_size=kernel_size)
```

```python
# compute P
P_nudged = preprocess.calculate_precipitation(Fx_left, Fx_right, Fy_bottom, Fy_top, E_nudged, dx, dy)
```

```python
coeffs_nudged = Coefficients(Fx_left, Fx_right, Fy_bottom, Fy_top, E_nudged, P_nudged, dx, dy)
```

```python
fig, ax = plt.subplots()
collection = plotting.pcolormesh(ax, coeffs_nudged.instability_heuristic, lon_axis, lat_axis, vmin=0.0)
fig.colorbar(collection, ax=ax, label="instability heuristic");
```

```python
print(
    f"Hot pixel identified at (i={i_hot}, j={j_hot}) "
    f"with instability heuristic = {coeffs_nudged.instability_heuristic[i_hot, j_hot]}"
)
print("E =", E_nudged[i_hot, j_hot])
print("P =", P_nudged[i_hot, j_hot])
```

solve the model again

```python
status = run(
    Fx_left,
    Fx_right,
    Fy_bottom,
    Fy_top,
    E_nudged,
    P_nudged,
    dx,
    dy,
    tol=1e-2,
)
```

```python
# converges
assert status["success"] is True
```

```python
# plot the final solution
rho = status["rho"]
fig, ax = plt.subplots()
collection = plotting.pcolormesh(ax, rho, lon_axis, lat_axis, vmin=0.0, vmax=0.8)
fig.colorbar(collection)
fig.suptitle("$\\rho$");
```

```python
# plot the convergence
deltas = status["deltas"]
fig, ax = plt.subplots()
ax.plot(deltas)
ax.set_title("Convergence")
ax.set_xlabel("Iteration");
```
